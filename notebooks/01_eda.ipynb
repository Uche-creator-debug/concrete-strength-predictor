```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# Load the data
df = pd.read_csv('../data/data.csv')
```

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:", df.isnull().sum().sum())
print("\nData types:\n", df.dtypes)
df.head()
```

```python
df.describe().round(2)
```

```python
df.duplicated().sum()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['csMPa'], kde=True, bins=30, color='blue')
plt.title('Distribution of Concrete Compressive Strength (csMPa)')
plt.xlabel('Compressive Strength (MPa)')
plt.show()
```

```python
plt.figure(figsize=(12, 8))
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f', linewidths=0.5)
plt.title('Feature Correlation with Compressive Strength')
plt.show()
```

```python
# Top features correlated with csMPa
correlations = df.corr()['csMPa'].sort_values(ascending=False)
print(correlations)
```
